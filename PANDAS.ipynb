```python
#DATA CLEANING USING PANDAS
import pandas as pd
data={
    "A":[1,2,3,None,5],
    "B":[None,2,3,4,5],
    "C":[1,2,None,4,5]

}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()

#USE DROPNA() TO REMOVE TO REMOVE ROWS WITH ANY OTHER VALUE
df_cleaned = df.dropna()
print(df_cleaned)
```

```python
#FIND THE MISSING VALUES
import pandas as pd
#define a dictionary with sample data which includes some missing values
data={
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,4,5]

}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()

#filling NaN values with 0
df.fillna(0,inplace=True)
print(df)
```

```python
#USE AGGREGATE FUNCTIONS TO FILL MISSING VALUES
import pandas as pd
#define a dictionary with simple data which includes some missing values
data={
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()
 #filling NaN vales with mean of each column
df.fillna(df.mean(),inplace=True)
print(df)
```

```python
#HOW TO HANDLE DUPLICATE VALUES
import pandas as pd
#sample data
data={
    'A':[1,2,3,3,4,4],
    'B':[4,6,6,7,8,8],
}
df=pd.DataFrame(data)
print("Original Data Frame:\n",df.to_string(index=False))
#detect duplicates
print("Duplicate Rows:\n",df[df.duplicated()].to_string(index=False))
#remove duplicates from column 'A'
df.drop_duplicates(subset=['A'],keep='first',inplace=True)
print("Data after removing duplicates based on column 'A':\n" ,df.to_string(index=False))
      
```

```python
#RENAME COLUMN NAMES TO MEANING NAMES
import pandas as pd

#sample data
data={
    'A':[25,34,56,78],
    'B':['Drishti','Sakshi','Ishmeet','Naman'],
    'C':[100000,50000,30000,200000]
}
df=pd.DataFrame(data)

# rename columns
df.rename(columns={'A':'Age','B':'Name','C':'Salary'},
          inplace=True)
print(df.to_string(index=False))
```
