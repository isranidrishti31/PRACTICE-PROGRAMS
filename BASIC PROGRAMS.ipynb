```python
#7 FIND THE SUM OF ALL GIVEN NUMBERS
num = int(input("Enter a number: "))

num = abs(num)
sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print("Sum of digits =", sum)
```

```python
#8 FIND THE FACTORIAL OF GIVEN NUMBER
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num + 1):
    factorial = factorial * i

print("Factorial =", factorial)
```

```python
#9 CHECK WHETHER THE GIVEN NUMBER IS PRIME OR NOT
num= int(input("Enter a number: "))

if num <= 1:
    print("The number is not prime")
else:
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print("The number is prime")
    else:
        print("The number is not prime")
```

```python
#12 REVERSE A GIVEN NUMBER
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number =", reverse)
```

```python
#13 CHECK WHETHER THE YEAR IS LEAP YEAR
year = int(input("Enter a year: "))

if year % 400 == 0:
    print("It is a leap year")
elif year % 100 == 0:
    print("It is not a leap year")
elif year % 4 == 0:
    print("It is a leap year")
else:
    print("It is not a leap year")
```

```python
#16 Student name, roll number, 3 subject marks, percentage, highest and lowest subject
name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")

maths = float(input("Enter marks in Maths: "))
science = float(input("Enter marks in Science: "))
english = float(input("Enter marks in English: "))

total = maths + science + english
percentage = total / 3

print("\nStudent Name:", name)
print("Roll Number:", roll_no)
print("Percentage:", percentage, "%")

# Finding highest marks
if maths >= science and maths >= english:
    highest = "Maths"
elif science >= maths and science >= english:
    highest = "Science"
else:
    highest = "English"

# Finding lowest marks
if maths <= science and maths <= english:
    lowest = "Maths"
elif science <= maths and science <= english:
    lowest = "Science"
else:
    lowest = "English"

print("Subject with Highest Marks:", highest)
print("Subject with Lowest Marks:", lowest)
```
