```python
#WRITE A PROGRAM THAT ACCEPTS THE THREE NUMBERS FROM USER AND DISPLAY ITS ADDITION USING FUNCTION
x=int(input("enter x"))
y=int(input("enter y"))
z=int(input("enter z"))

def cal_sum(x,y,z):
    sum=x+y+z
    return sum

s=cal_sum(x,y,z)
print(s)
```

```python
#CALCULATING SUM AND PRODUCT THROUGH A USER DEFINED FUNCTION CAL_SUM_PROD(_).
def cal_sum_prod(x,y,z):
    ss=x+y+z
    pp=x*y*z
    return pp,ss

a=int(input('Enter a:'))
b=int(input('Enter b:'))
c=int(input('Enter c:'))
s,p=cal_sum_prod(a,b,c)
print(s,p)
```

```python
#ACCEPT THE NUMBER FROM THE USER AND DISPLAY PRIME FACTORS OF THAT NUMBER
def prime_factors(n):
    for i in range (2,n+1):
        while n % i== 0:
            print(i,end= " " )
            while n%i ==0:
                n=n//i

num=int(input("enter number:"))
prime_factors(num)
```

```python
#WRITE A FUNCTION WHICH WILL PRINT ALL EVEN NUMBERS BETWEEN 1 TO 100
def even_numbers():
    for i in range (1,100):
        if i % 2 ==0:
            print(i,end=" ")

even_numbers()
```

```python
#RECURSION
def print_numbers(n):
    if n==0:   #Base condition
        return
    print_numbers(n-1)  #Recursive call
    print(n)

n=int(input("Enter n:"))
print_numbers(n)
```

```python
#RECURSION FACTORIAL
def factorial(n):
    if n==0 or n==1:
        return 1
    return n*factorial(n-1)
n=int(input("Enter a number:"))
print("Factorial=",factorial(n))

```

```python
#FIBONACCI SERIES
def fibonacci(n):
    if n<=1:
        return n
    return fibonacci(n-1)+ fibonacci(n-2)
n=int(input("Enter numner of terms:"))
for i in range(n):
    print(fibonacci(i),end="")
```
